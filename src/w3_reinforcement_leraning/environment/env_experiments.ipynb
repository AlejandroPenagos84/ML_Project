## All supervised model with classes

```python
from pathlib import Path
import sys
import pandas as pd

project_root = next(
    parent
    for parent in (Path.cwd(), *Path.cwd().parents)
    if (parent / "src" / "w3_reinforcement_leraning").is_dir()
)
sys.path.insert(0, str(project_root / "src"))

from w3_reinforcement_leraning.environment.customer_selector import CustomerSelector
from w3_reinforcement_leraning.environment.supervised_model.preprocessor_customer_data import (
    PreprocessorCustomerData,
)
from w3_reinforcement_leraning.environment.supervised_model.supervised_model import (
    SupervisedModel,
)

customer_selector = CustomerSelector()
customer_df = customer_selector.get_random_customer()

display(customer_df)
print(f"Row number: {customer_df.shape[0]}")


# Convierte la única fila del DataFrame a un diccionario para el preprocesador.
cliente = customer_df.iloc[0].to_dict()


preprocessor = PreprocessorCustomerData()
model = SupervisedModel()

customer_features = preprocessor.preprocess_customer(cliente)
churn_probability = model.predict_probability(customer_features)
prediction = model.predict(customer_features)

print(f"Predicción: {prediction}")
print(f"Probabilidad de churn: {churn_probability:.3f}")
```
