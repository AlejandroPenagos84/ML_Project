```python
from pathlib import Path
import sys

project_root = next(
    parent
    for parent in (Path.cwd(), *Path.cwd().parents)
    if (parent / 'workshops' / 'w3').is_dir()
)
sys.path.insert(0, str(project_root / 'workshops' / 'w3' / 'environment'))

from supervised_model.preprocessor_customer_data import PreprocessorCustomerData
from supervised_model.supervised_model import SupervisedModel
from supervised_model.customer_selector import CustomerSelector

customer_selector = CustomerSelector()
cliente = customer_selector.get_random_customer()

print('\n--- Member ---')
print({k: v for k, v in cliente.items() if k not in ('transactions', 'user_logs')})
print(f'Transactions: {len(cliente["transactions"])} records')
print(f'User logs   : {len(cliente["user_logs"])} records')

preprocessor = PreprocessorCustomerData()
model = SupervisedModel()

customer_features = preprocessor.preprocess_customer(cliente)
churn_probability = model.predict_probability(customer_features)
prediction = model.predict(customer_features)

print(f'\nPrediccion          : {prediction}')
print(f'Probabilidad churn  : {churn_probability:.3f}')
```
